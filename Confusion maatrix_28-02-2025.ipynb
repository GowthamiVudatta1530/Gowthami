```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
```

```
df = pd.read_csv('/content/sports_match_predictions.csv')
```

# **Business problem:-**Inaccurate sports match predictions lead to poor team decisions, betting losses, and low fan engagement

# **Business objective:-**A better prediction model can improve strategy, increase profits, and enhance audience experience

```
df
```

```
# Sample Actual and Predicted values
y_actual = ["Win", "Lose", "Win", "Lose", "Win", "Win", "Lose", "Lose"]
y_predicted = ["Win", "Win", "Lose", "Lose", "Win", "Lose", "Win", "Lose"]

# Compute confusion matrix
cm = confusion_matrix(y_actual, y_predicted, labels=["Win", "Lose"])

# Define class labels
labels = ["Win", "Lose"]

# Plot the confusion matrix as a heatmap
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=labels, yticklabels=labels)

# Labels, title, and formatting
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix for Sports Match Prediction")
plt.show()
```

# **True Positives (TP) = 2 → Model correctly predicted Win.
# False Negatives (FN) = 2 → Model wrongly predicted Lose when actual was Win.
# False Positives (FP) = 2 → Model wrongly predicted Win when actual was Lose.
# True Negatives (TN) = 2 → Model correctly predicted Lose.
**

```
# Compute confusion matrix
cm = confusion_matrix(y_actual, y_predicted, labels=["Win", "Lose"])

# Extract True Positives (TP), True Negatives (TN), False Positives (FP), and False Negatives (FN)
TP = cm[0, 0]  # True Positives (Win predicted as Win)
TN = cm[1, 1]  # True Negatives (Lose predicted as Lose)
FP = cm[0, 1]  # False Positives (Lose predicted as Win)
FN = cm[1, 0]  # False Negatives (Win predicted as Lose)

# Calculate accuracy
accuracy = (TP + TN) / (TP + TN + FP + FN)
print(f"Accuracy: {accuracy * 100}%")
```
